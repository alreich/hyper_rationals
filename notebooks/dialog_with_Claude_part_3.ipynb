## What was added

- **`is_zero_divisor()`** checks whether `y → x·y` is singular, using exact rational elimination, so it works at any rank and signature. It does not count 0.
  - Through rank 3 it agrees exactly with `is_null()`.
  - From rank 4 up it also finds zero divisors that are not null. Your known sedenion example, e1 + e10, is not null but is flagged.
  - It has no left/right argument. A value is a zero divisor on the left exactly when it is on the right, and exactly when its conjugate is.
  - The reason is an adjoint identity for the bilinear form, B(xy, z) = B(y, x̄z), which I checked in all 12 signatures I tried, up to rank 5. I did not write out a proof. The test suite includes that check.
- **`annihilator(kind="left"|"right")`** returns an exact basis of the values `y` with `x·y = 0` (or `y·x = 0`). It also serves as a witness: every returned vector is checked to be annihilated, and independent code checks the dimension against a rank computation (and SymPy's, when installed).

## What I found while building it

- **Sedenion annihilator size.** The annihilator of e1 + e10 has dimension 4, not 2 as I first wrote in a draft docstring. I caught it when running the example and fixed it.
- **Left and right annihilators.** They always have the same dimension, but they can be different subspaces. In the classical sedenion cases I sampled they coincided. In a split algebra such as signs (−1, −1, +1, −1) they often differ, e.g. for e0 + e4. The docs say only that.
- **Sample size.** The classical sedenion observation comes from about 360 sparse elements, not an exhaustive search, so it isn't a theorem.
