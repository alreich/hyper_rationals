### Block 3 -- quickstart (`.. code-block:: python`, under "Quickstart")

```
>>> from hyprat import Hy
>>> z = Hy('5/2', '-16/5')
>>> str(z)
'(5/2-16/5j)'

>>> q = Hy(Hy(1, 2), Hy(3, 4))   # a rational quaternion
>>> str(q)
'(1+2i+3j+4k)'

>>> i = Hy(Hy(0, 1), Hy(0, 0))
>>> j = Hy(Hy(0, 0), Hy(1, 0))
>>> str(i * j)
'(k)'

>>> Hy.from_array([1, 2, 3, 4])   # same quaternion, from a flat list
Hy(Hy('1', '2'), Hy('3', '4'))
```

**Note:** Block 3 is written in `index.rst` as doctest-style
(`>>> `) source -- it isn't valid to run directly as a normal code
cell as-is. Below is the same example as plain executable Python, so
you can actually run/edit it here; when you paste back into
`index.rst`, restore the `>>> ` / expected-output doctest formatting
shown above.


```
from hyprat import Hy

z = Hy('5/2', '-16/5')
print(str(z))                     # '(5/2-16/5j)'

q = Hy(Hy(1, 2), Hy(3, 4))        # a rational quaternion
print(str(q))                     # '(1+2i+3j+4k)'

i = Hy(Hy(0, 1), Hy(0, 0))
j = Hy(Hy(0, 0), Hy(1, 0))
print(str(i * j))                 # '(k)'

Hy.from_array([1, 2, 3, 4])       # same quaternion, from a flat list
```
