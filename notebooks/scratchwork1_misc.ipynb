## Sympy Interoperability

```python
from sympy import Quaternion

def hy_to_floats(hy):
    """Return all Hy elements as a list of floats."""
    return list(map(float, hy.to_array()))

def hy_to_sympy_quaternion(hy):
    return Quaternion.from_Matrix(hy_to_floats(hy))

def hy_from_sympy_quaternion(q):
    return Hy.from_array([q.a, q.b, q.c, q.d])
```

```python
h0 = Hy.random(2)
str(h0)
```

```python
q0 = hy_to_sympy_quaternion(h0)
q0
```
