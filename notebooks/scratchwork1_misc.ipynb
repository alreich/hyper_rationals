```python
u3 = Hy.units(3)
u3
```

```python
def find(unit, units):
    for key, val in units.items():
        if unit == val:
            return key
    else:
        return False
```

```python
o1 = u3['e4']
o2 = u3['e7']
find(o1 * o2, u3)
```

```python
u3x = dict(u3)

u3x[ "i"] = u3x.pop( "e1")
u3x["-i"] = u3x.pop("-e1")
u3x[ "j"] = u3x.pop( "e2")
u3x["-j"] = u3x.pop("-e2")
u3x[ "k"] = u3x.pop( "e3")
u3x["-k"] = u3x.pop("-e3")
u3x[ "l"] = u3x.pop( "e4")
u3x["-l"] = u3x.pop("-e4")

del u3x[ 'e5']
del u3x['-e5']
del u3x[ 'e6']
del u3x['-e6']
del u3x[ 'e7']
del u3x['-e7']

u3x
```

```python
for k1, v1 in u3x.items():
    for k2, v2 in u3x.items():
        prod = v1 * v2
        if not isinstance(prod, Hy):
            print("not a hy")
        if not find(prod, u3x):
            print(f"u3x[{k1 + k2}] == {prod!r}")
            break
```
