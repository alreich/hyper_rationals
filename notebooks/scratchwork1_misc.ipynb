```python
u3 = Hy.units(3)
u3
```

```python
def find(unit, units):
    for key, val in units.items():
        if unit == val:
            return key
    else:
        return False
```

```python
o1 = u3['e4']
o2 = u3['e7']
find(o1 * o2, u3)
```
