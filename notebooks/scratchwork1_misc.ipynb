## Sympy Interoperability

```python
from sympy import Quaternion

def hy_to_floats(hy):
    """Return all Hy elements as a list of floats."""
    return list(map(float, hy.to_array()))

def hy_to_sympy_quaternion(hy):
    return Quaternion.from_Matrix(hy_to_floats(hy))

def hy_from_sympy_quaternion(q):
    return Hy.from_array(list(map(float, [q.a, q.b, q.c, q.d])))
```

```python
# h0 = Hy.random(2)
h0 = Hy.from_string('(-1/5+5/2i-9/4j-2k)')
h0
```

```python
q0 = hy_to_sympy_quaternion(h0)
q0
```

```python
h0x = hy_from_sympy_quaternion(q0)
h0x
```

## Numpy Interoperability

```python
import numpy as np
import quaternion

# Create a single quaternion: w=1, x=0, y=0, z=0
q1 = np.quaternion(1, 0, 0, 0)
print(f"{q1 = }\n")

# Create an array of quaternions from a float array
float_data = np.array([[1.0, 0.0, 0.0, 0.0], [0.707, 0.0, 0.707, 0.0]])
q_array = quaternion.from_float_array(float_data)
print(f"{float_data = }")
print(f"{q_array = }\n")

# Fast vectorized math (multiplication, addition, logs, etc.)
q_product = q1 * q_array
conjugates = np.conjugate(q_array)
print(f"{q_product = }")
print(f"{conjugates = }")
```

```python
# Convert a quaternion array back to a standard float64 array instantly
# This creates a view of the exact same memory space
view_as_floats = quaternion.as_float_array(q_array) 
print(f"{view_as_floats = }")
```
