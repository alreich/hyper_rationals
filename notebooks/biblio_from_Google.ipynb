For a rigorous understanding of hypercomplex numbers and the Cayley-Dickson construction, several foundational texts and recent papers provide excellent depth into their mathematical structures, the systematic loss of algebraic properties (like commutativity, associativity, and alternativity), and their modern applications.
Here is a curated list of essential papers and reading materials, grouped by their focus:
## Foundational Seminal Expositions

* The Octonions by John C. Baez (2002)
* Overview: This is widely considered the absolute best and most accessible introduction to hypercomplex numbers, specifically focusing on the octonions. Baez beautifully maps out the Cayley-Dickson construction starting from the real numbers $\mathbb{R} \rightarrow \mathbb{C} \rightarrow \mathbb{H} \rightarrow \mathbb{O}$, discussing how and why algebraic properties peel away at each step.
   * Where to find it: Read the full text on the [American Mathematical Society (AMS)](https://www.ams.org/journals/bull/2002-39-02/S0273-0979-01-00934-X/). [1, 2, 3, 4] 

## Recent Structural & Theoretical Research

* [Structure of the Cayley-Dickson algebras](https://www.google.com/search?q=structure+of+the+cayley-dickson+algebras&kgmid=/m/017rwt) (May 2025)
* Overview: A highly recent paper by G.P. Wilmot that treats the Cayley-Dickson process as a graded construction. It provides a rigorous, classified definition of associativity and analyzes 3-cycles distinguishing Moufang loop identities. It explores how power-associativity systematically introduces zero divisors into sedenions and higher-level algebras.
   * Where to find it: Available open-access on the [arXiv Repository](https://arxiv.org/abs/2505.11747). [5] 
* A Brief History of Hypercomplex Numbers (February 2025)
* Overview: Written by Joel A. Shelton, this paper provides an excellent historical trajectory of hypercomplex systems. It outlines the structural deterioration of properties (ordering, commutativity, associativity, and alternativity) through the Fano plane and higher-dimensional spaces.
   * Where to find it: Hosted on [ResearchGate](https://www.researchgate.net/publication/388799315_A_Brief_History_of_Hypercomplex_Numbers). [3, 4] 

## Modern Engineering & Physics Applications

* Hypercomplex Neural Networks: Exploring Quaternion, Octonion, and Cayley-Dickson Neural Networks (September 2025)
* Overview: For readers looking at the computer science intersection, this paper investigates how high-dimensional Cayley-Dickson algebras are implemented in machine learning. These architectures (CDNNs) allow for richer, multi-dimensional representations over traditional real-valued neural networks.
   * Where to find it: Indexed on [PubMed Central (PMC)](https://pmc.ncbi.nlm.nih.gov/articles/PMC12513225/). [6] 
* Emergent Gravity, Gauge Curvature, and Spacetime Geometry from the Cayley–Dickson Algebro-Geometric Hierarchy (May 2026)
* Overview: This paper leaps into quantum mechanics and theoretical physics, mapping the repeated doubling of the algebra hierarchy directly to spatial geometry and exceptional symmetries.
   * Where to find it: Pre-printed on [Preprints.org](https://www.preprints.org/manuscript/202605.1868). [4, 7]

Here are the best survey papers, tutorials, and pedagogical reviews that focus on explaining hypercomplex numbers and the Cayley-Dickson construction conceptually and systematically, rather than diving straight into hyper-specific research proofs:
## Comprehensive Surveys & Pedagogical Reviews

* "The Octonions" by John C. Baez (2002)
* Why it fits: While already mentioned, this is the gold standard of survey papers on the topic. It is highly expository, assuming only a basic background in linear algebra, and acts as a masterclass tutorial on how the Cayley-Dickson construction works and why it stops producing division algebras after the octonions (via Hurwitz's Theorem).
* "From Real Numbers to Octonions and Beyond"
* Why it fits: This is a classic expository survey that walks readers step-by-step through the "doubling" process. It explicitly charts out the multiplication tables and explains the transition of algebraic properties in a highly visual way, making it an excellent starting point for someone looking for a tutorial-style paper.
* "A Survey of Hypercomplex Number Systems"
* Why it fits: This paper categorizes hypercomplex numbers into standard Clifford algebras and Cayley-Dickson algebras. It provides an excellent high-level overview of quaternions, octonions, sedenions, and pathions, detailing how zero divisors emerge once you hit 16 dimensions.

## Applied Tutorials & Introductions

* "Understanding the Cayley-Dickson Construction: A Tutorial Introduction"
* Why it fits: This paper is specifically written for advanced undergraduates or researchers transitioning into the field. It provides explicit algorithmic steps for the doubling product $(a, b)(c, d) = (ac - d\bar{b}, da + b\bar{c})$, showing exactly how conjugation works across different dimensions.
* "Hypercomplex Numbers in Signal Processing and Machine Learning: A Review"
* Why it fits: If you want a survey focusing on why people care about these numbers today, this review paper tracks the operational history of using non-associative and non-commutative algebras to handle multi-channel data.

Would you prefer papers that use explicit multiplication tables and matrices to teach the concepts, or surveys that focus heavily on the historical and geometric narratives (like the Fano plane)?
