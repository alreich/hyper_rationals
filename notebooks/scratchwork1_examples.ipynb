# Examples

```python
from hyprat import Hy
```

## Rational

```python
h0 = Hy(0)
h1 = Hy(1)
h1x = 2 * h1
```

```python
print(f"{h0 = }")
print(f"{h1 = }")
print(f"{h1x = }")
```

```python
print(f"{h1} + {h1x} = {h1 + h1x}\n")
```

## Complex

```python
h2 = Hy(0, 1)
h3 = Hy('1/2', '2/3')
h4 = Hy(3.2, -5.4)
h4x = 5 * h4
h5 = h3.inverse()
h6 = h4.conjugate()
```

```python
print(f"{h2 = }")
print(f"{h3 = }")
print(f"{h4 = }")
print(f"{h4x = }")
print(f"{h5 = }  # h3.inverse()")
print(f"{h6 = }  # h4.conjugate()")
```

```python
print(f"{h4} + {h3} = {h4 + h3}")
print(f"{h4} - {h3} = {h4 - h3}")
print(f"{h4} * {h3} = {h4 * h3}")
print(f"{h4} / {h3} = {h4 / h3}\n")

print(f"{h4x} / {h4} = {h4x / h4}\n")
```

## Quaternion

```python
h7 = Hy(h1, h2)
print(f"{h7 = }")  # Hy(h1 h2)
print(f"{str(h7) = }\n")

h8 = Hy(h3, h4)
print(f"{h8 = }")  # Hy(h3, h4)
print(f"{str(h8) = }\n")
```

```python
print(f"{h8} + {h7} = {h8 + h7}")
print(f"{h8} - {h7} = {h8 - h7}")
print(f"{h8} * {h7} = {h8 * h7}")
print(f"{h8} / {h7} = {h8 / h7}")
```
